**Something to visualize board for help**

```python
def view_board(board):
    for i in range(9):
        for j in range(9):
            if board[i][j]==0:
              print(".",end=" ")
            else:
              print(board[i][j],end=" ")

            if j==2 or j==5 or j==8:
              print("|",end=" ")

        if i==2 or i==5 or i==8:
          print()
          print("_ "*12)
        print()
```

**Function to help us find the first time 0 is found**

```python
def find_empty(board):
    for i in range(len(board)):
        for j in range(len(board[0])):
            if board[i][j] == 0:
                return (i, j)
    return None
```

**Now something to check if our entry is possibly correct or wrong**

```python
def is_valid(board,num,row,col):
  #Checking in row
  if num in board[row]:
    return False
  #cheking in colunm
  for i in range(9):
    if board[i][col] == num:
      return False
  #checking in the small matrix
  if 0<=row<=2:
    if 0<=col<=2:
      for i in range(3):
        for j in range(3):
          if board[i][j] == num:
            return False
    elif 3<=col<=5:
      for i in range(3):
        for j in range(3,6):
          if board[i][j] == num:
            return False
    else:
      for i in range(3):
        for j in range(6,9):
          if board[i][j] == num:
            return False
  elif 3<=row<=5:
    if 0<=col<=2:
      for i in range(3,6):
        for j in range(3):
          if board[i][j] == num:
            return False
    elif 3<=col<=5:
      for i in range(3,6):
        for j in range(3,6):
          if board[i][j] == num:
            return False
    else:
      for i in range(3,6):
        for j in range(6,9):
          if board[i][j] == num:
            return False
  else:
    if 0<=col<=2:
      for i in range(6,9):
        for j in range(3):
          if board[i][j] == num:
            return False
    elif 3<=col<=5:
      for i in range(6,9):
        for j in range(3,6):
          if board[i][j] == num:
            return False
    else:
      for i in range(6,9):
        for j in range(6,9):
          if board[i][j] == num:
            return False

  return True
```

**Seeing if input is valid**

```python
def invalid_input(board):
  for i in range(9):
    for j in range(9):
      val=board[i][j]
      if val !=0:
        board[i][j]=0
        if not is_valid(board,val,i,j):
            return True
        else:
          board[i][j]=val

  return False
```

**Function to enter something in the empty cell we find**

```python
def solve(board):
  res=find_empty(board)
  x=invalid_input(board)
  if x:
    return "Wrong input"
  if res is None:
    return True
  row,col=res
  for i in range(1,10):
    if is_valid(board,i,row,col):
      board[row][col]=i
      if solve(board):
        return True
      else:
        board[row][col]=0

  return False
```

**Taking input here**

```python
puzzle = []


with open("puzzle.txt", "r") as file:
    for line in file:

        clean_line = line.strip()
        if clean_line:
            row = [int(digit) for digit in clean_line]
            puzzle.append(row)

print("Puzzle loaded from file:")
view_board(puzzle)
```

**Let us solve!!**

```python
output=solve(puzzle)
if output=="Wrong input":
  print("Wrong input")
else:
  view_board(puzzle)
```
